## 6. Independent re-implementation (sanity check)

Thirty lines that recompute Min-K% Prob from scratch, with its own progress bar.
If this agrees with the repository's number, the sign convention noted as issue (iii)
in Section 4.2 has been read correctly; if it comes out at `1 - AUC`, the ROC curve
is inverted. This cell also works standalone if cell 4 ever fails.

```python
import numpy as np, torch
from transformers import AutoTokenizer, AutoModelForCausalLM
from sklearn.metrics import roc_auc_score
from tqdm.auto import tqdm

device = 'cuda' if torch.cuda.is_available() else 'cpu'
tok = AutoTokenizer.from_pretrained(TARGET)
model = AutoModelForCausalLM.from_pretrained(TARGET).to(device).eval()

def token_logprobs(text):
    ids = torch.tensor(tok.encode(text)).unsqueeze(0).to(device)
    with torch.no_grad():
        logits = model(ids).logits
    lp = torch.log_softmax(logits, dim=-1)[0]
    return [lp[i, t].item() for i, t in enumerate(ids[0][1:])]

def min_k_prob(text, k=0.2):
    lps = token_logprobs(text)
    n = max(1, int(len(lps) * k))
    return float(np.mean(sorted(lps)[:n]))   # higher (less negative) => member

scores = [min_k_prob(ex['input']) for ex in tqdm(d32, desc='scoring')]
labels = [ex['label'] for ex in d32]
print(f'independent Min-20% AUC = {roc_auc_score(labels, scores):.3f}')
try:
    print(f'repository  Min-20% AUC = {mink20:.3f}')
except NameError:
    print('(repository run not available — use the number above)')
```
